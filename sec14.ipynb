### Peer Presentations

Form a group of four students. For each round, there will be a **presenter** and three **reviewers**.

Reviewers should give feedback on:

- The presentation
  - Are the slides easy to follow?
  - Are the intentions and decisions clear?

- The demo
  - Does it solve the problem the presenter described?
  - What could they improve for the final submission?

----

### How each round works

1. (5 min) The presenter gives their presentation and a short demo of their project.

2. (10 mins) The presenter lets the reviewers test and explore the project on the presenter's computer. As the reviewers explore, they give feedback on the presentation and code using the Ladder of Feedback. The presenter takes notes in the section below.

3. Swap roles and repeat.

----

### The Ladder of Feedback

1.  **Clarifying:** Ask questions to understand the intent of each part of the project.
    - “What were you trying to accomplish when you…?”
    - “Can you walk me through what happens when…?”

2.  **Valuing:** Express what you like about the work using specific details.
    - “I really liked how you… because…”
    - “One thing that stood out to me was…”

3.  **Offering concerns:** Raise areas for improvement. Hold off on suggestions.
    - “I wonder if… might cause a problem when…”
    - “I was a little confused by… because…”
    
4.  **Suggesting:** Propose one or two paths forward to address the concerns.
    - “Have you considered trying…?”
    - “One possible next step could be…”
